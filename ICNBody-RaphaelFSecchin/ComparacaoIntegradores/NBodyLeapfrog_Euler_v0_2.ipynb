```python
espacamento = 10
passos = 15
g = 1
epsilon = 5
dt = 0.1
threadsperblock = 32
numCorpos = 2**12
corposPorEixo1D = numCorpos
corposPorEixo2D = int(math.sqrt(numCorpos))
corposPorEixo3D = int(math.cbrt(numCorpos))
frequenciaSnapshots = 15
```

A função calculaForcas calcula o valor das forças do sistema utilizando a fórmula:

$$ F_i = ∑_{\substack{j = 1 \\ j \neq i}}^{N} \frac{G m_{i} m_{j}}{|\vec{r}_{ij}|^{2}}$$

A função calculaEnergiaK calcula o valor da energia cinética utilizando a fórmula:

$$ K = ∑^{N}_{i = 1}\frac{m_{i}|\vec{v}_{i}|^{2}}{2}$$

A função calculaEnergiaU calcula o valor da energia cinética utilizando a fórmula:

$$ U = ∑^{N}_{i = 1}∑^{N}_{\substack{j = 1 \\ j \neq i}}\frac{G m_{i} m_{j}}{|r_{ij}|}$$

```python
x = []
v = dados["vx"].to_numpy().tolist()
m = dados["massa"].to_numpy().tolist()
forca = [0.0] * numCorpos
xTemp = 0

for i in range(corposPorEixo1D):
  xTemp += espacamento
  x.append(xTemp)
```

```python
def calculaForcas(numCorpos, x, m, g, epsilon, forca):
  for i in range(numCorpos):
    for j in range(i + 1, numCorpos):

      dx = x[j] - x[i]
      dist = (dx*dx + epsilon*epsilon)

      invDist = dist**(-0.5)
      invDist = invDist * invDist * invDist

      forcaAtual = (g * m[i] * m[j]) * invDist

      forca[i] += forcaAtual * dx
      forca[j] += -forcaAtual * dx
```

```python
def movimentaCorpo(dt, v, x, m, numCorpos, g, epsilon, forca):
  for k in range(numCorpos):
    forca[k] = 0.0

  calculaForcas(numCorpos, x, m, g, epsilon, forca)
  for j in range(numCorpos):

    acel = forca[j] / m[j]

    v[j] = v[j] + acel * dt

    x[j] = x[j] + v[j] * dt
```

```python
def calculaEnergiaK(m, v, numCorpos):
  k = 0.0
  for i in range(numCorpos):
    k += 0.5 * m[i] * v[i]*v[i]
  return k
```

```python
def calculaEnergiaU(numCorpos, x, m, g, epsilon):
  u = 0.0
  for i in range(numCorpos):
    for j in range(i + 1, numCorpos):
      dx = x[j] - x[i]
      dist = (dx*dx + epsilon*epsilon)**0.5

      u += - (g * m[i] * m[j]) / dist

  return u
```

Tempo de Execução:

```python
%time kInicial = calculaEnergiaK(m, v, numCorpos)
```

```python
%time uInicial = calculaEnergiaU(numCorpos, x, m, g, epsilon)
```

```python
snapshots = np.empty([len(range(0, passos, frequenciaSnapshots)) * numCorpos, 8], dtype=np.float64)
contSnap = 1
```

```python
%%time
for i in range(passos):
  if(i % frequenciaSnapshots == 0):
      inicio = contSnap * numCorpos - numCorpos
      fim = contSnap * numCorpos
      contSnap += 1
      snapshots[inicio:fim, 0:8] = np.array([x, np.full(numCorpos, 0), np.full(numCorpos, 0), v, np.full(numCorpos, 0),
                                             np.full(numCorpos, 0), m, np.full(numCorpos, i)]).transpose()

  movimentaCorpo(dt, v, x, m, numCorpos, g, epsilon, forca)
```

```python
df = pd.DataFrame(
    snapshots,
    columns=['x', 'y', 'z', 'vx', 'vy', 'vz', 'm', 'passo']
)

salvarSnapshot(df, biblioteca="PythonPuro", dimensao="D1")
```

```python
%time kFinal = calculaEnergiaK(m, v, numCorpos)
```

```python
%time uFinal = calculaEnergiaU(numCorpos, x, m, g, epsilon)
```

```python
def calcularForcas(np_x, np_m, G, epsilon):
    dx = np_x[None, :] - np_x[:, None]

    dist = np.sqrt(dx**2 + epsilon**2)

    np.fill_diagonal(dist, np.inf)

    F = G * (np_m[:, None] * np_m[None, :]) / (dist**2)

    fx = F * (dx / dist)

    fx_total = np.sum(fx, axis=1)

    return fx_total
```

```python
def atualizarPosicao(np_x, np_vx, np_m, dt, G, epsilon):
    fx = calcularForcas(np_x, np_m, G, epsilon)

    ax = fx / np_m

    np_vx += ax * dt

    np_x += np_vx * dt
```

```python
x = []
y = []
z = []
vx = dados["vx"].to_numpy().tolist()
vy = dados["vy"].to_numpy().tolist()
vz = dados["vz"].to_numpy().tolist()
m = dados["massa"].to_numpy().tolist()
xTemp = 0

forcaX = [0.0] * numCorpos
forcaY = [0.0] * numCorpos
forcaZ = [0.0] * numCorpos

for i in range(corposPorEixo3D):
  xTemp += espacamento
  yTemp = 0
  for j in range(corposPorEixo3D):
    yTemp += espacamento
    zTemp = 0
    for k in range(corposPorEixo3D):
      zTemp += espacamento
      x.append(xTemp)
      y.append(yTemp)
      z.append(zTemp)
```

```python
np_x = np.array(x, dtype=np.float64)
np_y = np.array(y, dtype=np.float64)
np_z = np.array(z, dtype=np.float64)
np_m = np.array(m, dtype=np.float64)
np_vx = np.array(vx, dtype=np.float64)
np_vy = np.array(vy, dtype=np.float64)
np_vz = np.array(vz, dtype=np.float64)
```

```python

def calcularForcas(np_x, np_y, np_z, np_m, G, epsilon):
    dx = np_x[None, :] - np_x[:, None]
    dy = np_y[None, :] - np_y[:, None]
    dz = np_z[None, :] - np_z[:, None]

    dist = np.sqrt(dx**2 + dy**2 + dz**2 + epsilon**2)

    np.fill_diagonal(dist, np.inf)

    F = G * (np_m[:, None] * np_m[None, :]) / (dist**2)

    fx = F * (dx / dist)
    fy = F * (dy / dist)
    fz = F * (dz / dist)

    fx_total = np.sum(fx, axis=1)
    fy_total = np.sum(fy, axis=1)
    fz_total = np.sum(fz, axis=1)

    return fx_total, fy_total, fz_total
```

```python

def atualizarPosicao(np_x, np_y, np_z, np_vx, np_vy, np_vz, np_m, dt, G, epsilon):
    fx, fy, fz = calcularForcas(np_x, np_y, np_z, np_m, G, epsilon)

    ax = fx / np_m
    ay = fy / np_m
    az = fz / np_m

    np_vx += ax * dt
    np_vy += ay * dt
    np_vz += az * dt

    np_x += np_vx * dt
    np_y += np_vy * dt
    np_z += np_vz * dt
```

```python
def calcularEnergiaCinetica(np_vx, np_vy, np_vz, np_m):
    return 0.5 * np.sum(np_m * (np_vx**2 + np_vy**2 + np_vz**2))
```

```python

def calcularEnergiaPotencial(np_x, np_y, np_z, np_m, G, epsilon):
    dx = np_x[None, :] - np_x[: ,None]
    dy = np_y[None, :] - np_y[: ,None]
    dz = np_z[None, :] - np_z[: ,None]

    dist = np.sqrt(dx**2 + dy**2 + dz**2 + epsilon**2)

    np.fill_diagonal(dist, np.inf)

    m_prod = np_m[:, None] * np_m[None, :]

    U = G * np.sum(m_prod / dist) / 2

    return U
```

```python
@cuda.jit
def calculaEnergiaK(mD, vxD, vyD, vzD, numCorpos, kD):
  i = cuda.grid(1)
  kD[i] = 0.5 * mD[i] * (vxD[i]*vxD[i] + vyD[i]*vyD[i] + vzD[i]*vzD[i])
```

```python
@cuda.jit
def calculaEnergiaU(numCorpos, xD, yD, zD, mD, g, epsilon, uD):
  i = cuda.grid(1)
  ep2 = epsilon*epsilon
  if(i < numCorpos):
    u = 0.0
    for j in range(i + 1, numCorpos):
      dx = xD[j] - xD[i]
      dy = yD[j] - yD[i]
      dz = zD[j] - zD[i]
      dist = (dx*dx + dy*dy + dz*dz + ep2)**0.5

      u += - (g * mD[i] * mD[j]) / dist
    uD[i] = (u + (g * mD[i] * mD[i]) / epsilon) / 2
```

```python
%time calculaEnergiaK[blockspergrid, threadsperblock](mD, vxD, vyD, vzD, numCorpos, kD)
kInicial = sum(kD)
```

```python
%time calculaEnergiaU[blockspergrid, threadsperblock](numCorpos, xD, yD, zD, mD, g, epsilon, uD)
uInicial = sum(uD)
```

```python
snapshots = np.empty([len(range(0, passos, frequenciaSnapshots)) * numCorpos, 8], dtype=np.float64)
contSnap = 1
```

```python
%%time
for i in range(passos):
  if(i % frequenciaSnapshots == 0):
      inicio = contSnap * numCorpos - numCorpos
      fim = contSnap * numCorpos
      contSnap += 1
      snapshots[inicio:fim, 0:8] = np.array([xD, yD, zD, vxD, vyD,
                                             vzD, mD, np.full(numCorpos, i)]).transpose()

  movimentaCorpo[blockspergrid, threadsperblock](dt, vxD, vyD, vzD, xD, yD, zD, mD, numCorpos, g, epsilon, forcaXD, forcaYD, forcaZD)
```

```python
def calculaForcas(numCorpos, x, m, g, epsilon, forca):
  for i in range(numCorpos):
    for j in range(i + 1, numCorpos):

      dx = x[j] - x[i]
      dist = (dx*dx + epsilon*epsilon)

      invDist = dist**(-0.5)
      invDist = invDist * invDist * invDist

      forcaAtual = (g * m[i] * m[j]) * invDist

      forca[i] += forcaAtual * dx
      forca[j] += -forcaAtual * dx
```

```python
def movimentaCorpo(dt, v, x, m, numCorpos, g, epsilon, forca):
  for k in range(numCorpos):
    forca[k] = 0.0

  calculaForcas(numCorpos, x, m, g, epsilon, forca)

  for j in range(numCorpos):
      a = forca[j] / m[j]
      v[j] += 0.5 * dt * a
      x[j] += v[j] * dt

  calculaForcas(numCorpos, x, m, g, epsilon, forca)

  for j in range(numCorpos):
      a = forca[j] / m[j]
      v[j] += 0.5 * dt * a
```

```python
def calculaEnergiaU(numCorpos, x, m, g, epsilon):
  u = 0.0
  for i in range(numCorpos):
    for j in range(i + 1, numCorpos):
      dx = x[j] - x[i]
      dist = (dx*dx + epsilon*epsilon)**0.5

      u += - (g * m[i] * m[j]) / dist

  return u
```

```python
@njit(cache=True)
def movimentaCorpo(dt, v, x, m, numCorpos, g, epsilon, forca):
  for k in range(numCorpos):
    forca[k] = 0.0

  calculaForcas(numCorpos, x, m, g, epsilon, forca)

  for j in range(numCorpos):
      a = forca[j] / m[j]
      v[j] += 0.5 * dt * a
      x[j] += v[j] * dt

  calculaForcas(numCorpos, x, m, g, epsilon, forca)

  for j in range(numCorpos):
      a = forca[j] / m[j]
      v[j] += 0.5 * dt * a
```

```python
@njit(cache=True)
def calculaEnergiaU(numCorpos, x, m, g, epsilon):
  u = 0.0
  for i in range(numCorpos):
    for j in range(i + 1, numCorpos):
      dx = x[j] - x[i]
      dist = (dx*dx + epsilon*epsilon)**0.5

      u += - (g * m[i] * m[j]) / dist

  return u
```

```python
@cuda.jit(device=True)
def calculaForcas(numCorpos, xD, mD, g, epsilon, forcaD, i):
  xLocal = cuda.shared.array(threadsperblock, cuda.float64)
  mLocal = cuda.shared.array(threadsperblock, cuda.float64)

  xDAtual = xD[i]
  mDAtual = mD[i]

  lx = cuda.threadIdx.x
  ep2 = epsilon*epsilon

  forcaLocal = 0.0
  for j in range(numCorpos // threadsperblock):
    xLocal[lx] = xD[lx + (threadsperblock * j)]
    mLocal[lx] = mD[lx + (threadsperblock * j)]

    cuda.syncthreads()

    for k in range(threadsperblock):
      dx = xLocal[k] - xDAtual
      dist = (dx*dx + ep2)

      invDist = dist**(-0.5)
      invDist = invDist * invDist * invDist

      forcaAtual = (g * mDAtual * mLocal[k]) * invDist
      forcaLocal += forcaAtual * dx

    cuda.syncthreads()

  forcaD[i] = forcaLocal
```

```python
@cuda.jit
def movimentaCorpo(dt, vD, xD, mD, numCorpos, g, epsilon, forcaD):
    i = cuda.grid(1)

    if(i < numCorpos):
      forcaD[i] = 0.0
      calculaForcas(numCorpos, xD, mD, g, epsilon, forcaD, i)

      acel = forcaD[i] / mD[i]
      vD[i] = vD[i] + acel * dt * 0.5
      xD[i] = xD[i] + vD[i] * dt

      calculaForcas(numCorpos, xD, mD, g, epsilon, forcaD, i)

      acel = forcaD[i] / mD[i]
      vD[i] = vD[i] + acel * dt * 0.5
```

```python
@cuda.jit(device=True)
def calculaForcas(numCorpos, xD, yD, mD, g, epsilon, forcaXD, forcaYD, i):
  xLocal = cuda.shared.array(threadsperblock, cuda.float64)
  yLocal = cuda.shared.array(threadsperblock, cuda.float64)
  mLocal = cuda.shared.array(threadsperblock, cuda.float64)

  xDAtual = xD[i]
  yDAtual = yD[i]
  mDAtual = mD[i]

  lx = cuda.threadIdx.x

  forcaLocalX = 0.0
  forcaLocalY = 0.0

  ep2 = epsilon*epsilon
  for j in range(numCorpos // threadsperblock):
    xLocal[lx] = xD[lx + (threadsperblock * j)]
    yLocal[lx] = yD[lx + (threadsperblock * j)]
    mLocal[lx] = mD[lx + (threadsperblock * j)]

    cuda.syncthreads()
    for k in range(threadsperblock):
      dx = xLocal[k] - xDAtual
      dy = yLocal[k] - yDAtual
      dist = (dx*dx + dy*dy + ep2)

      invDist = dist**(-0.5)
      invDist = invDist * invDist * invDist

      forcaAtual = (g * mDAtual * mLocal[k]) * invDist
      forcaLocalX += forcaAtual * dx
      forcaLocalY += forcaAtual * dy

    cuda.syncthreads()

  forcaXD[i] = forcaLocalX
  forcaYD[i] = forcaLocalY
```

```python
@cuda.jit
def movimentaCorpo(dt, vxD, vyD, xD, yD, mD, numCorpos, g, epsilon, forcaXD, forcaYD):
    i = cuda.grid(1)

    if(i < numCorpos):
      forcaXD[i] = 0.0
      forcaYD[i] = 0.0

      calculaForcas(numCorpos, xD, yD, mD, g, epsilon, forcaXD, forcaYD, i)

      acelX = forcaXD[i] / mD[i]
      acelY = forcaYD[i] / mD[i]

      vxD[i] = vxD[i] + acelX * dt * 0.5
      vyD[i] = vyD[i] + acelY * dt * 0.5

      xD[i] = xD[i] + vxD[i] * dt
      yD[i] = yD[i] + vyD[i] * dt

      calculaForcas(numCorpos, xD, yD, mD, g, epsilon, forcaXD, forcaYD, i)

      acelX = forcaXD[i] / mD[i]
      acelY = forcaYD[i] / mD[i]

      vxD[i] = vxD[i] + acelX * dt * 0.5
      vyD[i] = vyD[i] + acelY * dt * 0.5
```

```python
@cuda.jit
def calculaEnergiaK(mD, vxD, vyD, numCorpos, kD):
  i = cuda.grid(1)
  kD[i] = 0.5 * mD[i] * (vxD[i]*vxD[i] + vyD[i]*vyD[i])
```

```python
@cuda.jit
def calculaEnergiaU(numCorpos, xD, yD, mD, g, epsilon, uD):
  i = cuda.grid(1)
  ep2 = epsilon*epsilon

  if(i < numCorpos):
    u = 0.0

    for j in range(numCorpos):
      dx = xD[j] - xD[i]
      dy = yD[j] - yD[i]
      dist = (dx*dx + dy*dy + ep2)**0.5

      u += - (g * mD[i] * mD[j]) / dist

    uD[i] = (u + (g * mD[i] * mD[i]) / epsilon) / 2
```

```python
xH = []
yH = []
zH = []
vxH = dados["vx"].to_numpy().copy()
vyH = dados["vy"].to_numpy().copy()
vzH = dados["vz"].to_numpy().copy()
mH = dados["massa"].to_numpy().copy()
kH = [0.0] * numCorpos
uH = [0.0] * numCorpos
xTemp = 0

forcaXH = [0.0] * numCorpos
forcaYH = [0.0] * numCorpos
forcaZH = [0.0] * numCorpos

for i in range(corposPorEixo3D):
  xTemp += espacamento
  yTemp = 0
  for j in range(corposPorEixo3D):
    yTemp += espacamento
    zTemp = 0
    for k in range(corposPorEixo3D):
      zTemp += espacamento
      xH.append(xTemp)
      yH.append(yTemp)
      zH.append(zTemp)

xH = np.array(xH, dtype=np.float64)
yH = np.array(yH, dtype=np.float64)
zH = np.array(zH, dtype=np.float64)
mH = np.array(mH, dtype=np.float64)
vxH = np.array(vxH, dtype=np.float64)
vyH = np.array(vyH, dtype=np.float64)
vzH = np.array(vzH, dtype=np.float64)

forcaXH = np.array(forcaXH, dtype=np.float64)
forcaYH = np.array(forcaYH, dtype=np.float64)
forcaZH = np.array(forcaZH, dtype=np.float64)
kH = np.array(kH, dtype=np.float64)
uH = np.array(uH, dtype=np.float64)

xD = cuda.to_device(xH)
yD = cuda.to_device(yH)
zD = cuda.to_device(zH)
vxD = cuda.to_device(vxH)
vyD = cuda.to_device(vyH)
vzD = cuda.to_device(vzH)
mD = cuda.to_device(mH)
kD = cuda.to_device(kH)
uD = cuda.to_device(uH)
forcaXD = cuda.to_device(forcaXH)
forcaYD = cuda.to_device(forcaYH)
forcaZD = cuda.to_device(forcaZH)
```

```python
@cuda.jit(device=True)
def calculaForcas(numCorpos, xD, yD, zD, mD, g, epsilon, forcaXD, forcaYD, forcaZD, i):
  xLocal = cuda.shared.array(threadsperblock, cuda.float64)
  yLocal = cuda.shared.array(threadsperblock, cuda.float64)
  zLocal = cuda.shared.array(threadsperblock, cuda.float64)
  mLocal = cuda.shared.array(threadsperblock, cuda.float64)

  xDAtual = xD[i]
  yDAtual = yD[i]
  zDAtual = zD[i]
  mDAtual = mD[i]

  lx = cuda.threadIdx.x

  forcaLocalX = 0.0
  forcaLocalY = 0.0
  forcaLocalZ = 0.0

  ep2 = epsilon*epsilon
  for j in range(numCorpos // threadsperblock):
    xLocal[lx] = xD[lx + (threadsperblock * j)]
    yLocal[lx] = yD[lx + (threadsperblock * j)]
    zLocal[lx] = zD[lx + (threadsperblock * j)]
    mLocal[lx] = mD[lx + (threadsperblock * j)]

    cuda.syncthreads()
    for k in range(threadsperblock):
      dx = xLocal[k] - xDAtual
      dy = yLocal[k] - yDAtual
      dz = zLocal[k] - zDAtual
      dist = (dx*dx + dy*dy + dz*dz + ep2)

      invDist = dist**(-0.5)
      invDist = invDist * invDist * invDist

      forcaAtual = (g * mDAtual * mLocal[k]) * invDist
      forcaLocalX += forcaAtual * dx
      forcaLocalY += forcaAtual * dy
      forcaLocalZ += forcaAtual * dz

    cuda.syncthreads()

  forcaXD[i] = forcaLocalX
  forcaYD[i] = forcaLocalY
  forcaZD[i] = forcaLocalZ
```

```python
@cuda.jit
def movimentaCorpo(dt, vxD, vyD, vzD, xD, yD, zD, mD, numCorpos, g, epsilon, forcaXD, forcaYD, forcaZD):
    i = cuda.grid(1)

    if(i < numCorpos):
      forcaXD[i] = 0.0
      forcaYD[i] = 0.0
      forcaZD[i] = 0.0

      calculaForcas(numCorpos, xD, yD, zD, mD, g, epsilon, forcaXD, forcaYD, forcaZD, i)

      acelX = forcaXD[i] / mD[i]
      acelY = forcaYD[i] / mD[i]
      acelZ = forcaZD[i] / mD[i]

      vxD[i] = vxD[i] + acelX * dt * 0.5
      vyD[i] = vyD[i] + acelY * dt * 0.5
      vzD[i] = vzD[i] + acelZ * dt * 0.5

      xD[i] = xD[i] + vxD[i] * dt
      yD[i] = yD[i] + vyD[i] * dt
      zD[i] = zD[i] + vzD[i] * dt

      calculaForcas(numCorpos, xD, yD, zD, mD, g, epsilon, forcaXD, forcaYD, forcaZD, i)

      acelX = forcaXD[i] / mD[i]
      acelY = forcaYD[i] / mD[i]
      acelZ = forcaZD[i] / mD[i]

      vxD[i] = vxD[i] + acelX * dt * 0.5
      vyD[i] = vyD[i] + acelY * dt * 0.5
      vzD[i] = vzD[i] + acelZ * dt * 0.5
```
